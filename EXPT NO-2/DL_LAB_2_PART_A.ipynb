 # PART A: VISUALIZATION OF ACTIVATION FUNCTIONS

```python
# Import necessary libraries

import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
print('NAME:  AKHIL C\nROLL NO:  24BAD007')
```

```python
# Create input values from -10 to 10
x = np.linspace(-10, 10, 200)

# Display the first 10 values
print(x[:10])
```

# SIGMOID ACTIVATION FUNCTION

```python
# Sigmoid Activation Function
y_sigmoid = tf.keras.activations.sigmoid(x)

# Display the first 10 output values
print(y_sigmoid[:10])
```

# TANH ACTIVATION FUNCTION

```python
# Tanh Activation Function
y_tanh = tf.keras.activations.tanh(x)

# Display the first 10 output values
print(y_tanh[:10])
```

# ReLU ACTIVATION FUNCTION

```python
# ReLU Activation Function
y_relu = tf.keras.activations.relu(x)

# Display the first 10 output values
print(y_relu[:10])
```

```python
# Plot all activation functions

plt.figure(figsize=(12,5))

# Sigmoid
plt.subplot(1,3,1)
plt.plot(x, y_sigmoid, color='blue')
plt.title("Sigmoid")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)

# Tanh
plt.subplot(1,3,2)
plt.plot(x, y_tanh, color='green')
plt.title("Tanh")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)

# ReLU
plt.subplot(1,3,3)
plt.plot(x, y_relu, color='red')
plt.title("ReLU")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)

plt.tight_layout()
plt.show()
```
